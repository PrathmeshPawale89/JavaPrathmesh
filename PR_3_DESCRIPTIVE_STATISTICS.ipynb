```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/PrathmeshPawale89/my-website/refs/heads/main/StudentsPerformance.csv"
df = pd.read_csv(url)

df.head()
```

```python
grouped_stats = df.groupby('gender')[['math score', 'reading score', 'writing score']].agg(
    ['mean', 'median', 'min', 'max', 'std']
)

grouped_stats
```

```python
math_scores_list = df.groupby('gender')['math score'].apply(list)

math_scores_list
```

```python
sns.boxplot(x='gender', y='math score', data=df)
plt.title("Math Score Distribution by Gender")
plt.show()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
iris_stats = iris.groupby('species').agg(['mean', 'std', 'min', 'max'])

iris_stats
```

```python
iris_percentiles = iris.groupby('species').quantile([0.25, 0.5, 0.75])

iris_percentiles
```

```python
sns.boxplot(x='species', y='sepal_length', data=iris)
plt.title("Sepal Length Distribution by Species")
plt.show()
```
